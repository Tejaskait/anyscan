```
import yaml
cfg = yaml.safe_load(open('/content/dataset/data.yaml'))
cfg['path'] = '/content/dataset'
yaml.safe_dump(cfg, open('/content/dataset/data.yaml','w'), sort_keys=False)
print(cfg)
```

```
from ultralytics import YOLO
model = YOLO('yolo11n.pt')
model.train(
    data='/content/dataset/data.yaml',
    epochs=100, imgsz=640, batch=16, patience=20,
    project='/content/drive/MyDrive/anyscan/runs', name='round1',
)
```

```
metrics = YOLO('/content/drive/MyDrive/anyscan/runs/round1/weights/best.pt').val()
print('mAP50', metrics.box.map50, 'P', metrics.box.mp, 'R', metrics.box.mr)
```
